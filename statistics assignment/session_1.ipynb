task 1.

Open Microsoft Excel or Google Sheets, create a column named 'Order Amount', and enter 20 random numbers between 100 and 1000 to simulate Zomato order values for a day.

```python
import pandas as pd
import random
```

```python
random.seed(10)

order_amounts = [random.randint(100, 1000) for i in range(20)]

df = pd.DataFrame({
    'Order Amount': order_amounts
})

print(df)

df.to_excel('Zomato_Order_Statistics.xlsx', index=False)

print("Excel file created successfully!")
```

task 2.
From your 'Order Amount' data, calculate the mean, median, and mode using Excel or Sheets functions (AVERAGE, MEDIAN, MODE) and write the results below your data.


```python
random.seed(10)

order_amounts = [random.randint(100, 1000) for i in range(20)]

df = pd.DataFrame({
    'Order Amount': order_amounts
})

print("ZOMATO ORDER DATA")
display(df)


mean_value = df['Order Amount'].mean()

median_value = df['Order Amount'].median()

mode_value = df['Order Amount'].mode()

print("STATISTICAL RESULTS")

print("Mean:", mean_value)

print("Median:", median_value)

if len(mode_value) == 0:
    print("Mode: No Mode")
else:
    print("Mode:", list(mode_value))


df.to_csv("Zomato_Order_Statistics.csv", index=False)

print("\nFile saved successfully!")
```

task 3.
Select the first 5 values from your 'Order Amount' column as a sample. Calculate the sample mean and compare it to the population mean you found earlier.<br><br><em><strong>Hint:</strong> Use the AVERAGE function for both calculations and observe the difference between sample and population means.</em>


```python
population_mean = df['Order Amount'].mean()

sample = df['Order Amount'].head(5)

sample_mean = sample.mean()

print("First 5 Orders (Sample):")
print(sample)

print("\nPopulation Mean:", population_mean)
print("Sample Mean:", sample_mean)

difference = abs(population_mean - sample_mean)

print("Difference:", difference)
```

task 4.
Write a short paragraph explaining the difference between 'population' and 'sample' using your Zomato order data as an example. Mention which one is more practical to use for a food delivery company and why.


task 4 (answer)

Population means the complete collection of data. In our Zomato example, all 20 order amounts represent the population. A sample is a small part selected from the population. For example, the first 5 order amounts represent a sample. For a food delivery company, using a sample is more practical because the company may receive thousands of orders every day. Analysing a sample saves time and makes data analysis easier. It also helps the company understand customer ordering patterns.
