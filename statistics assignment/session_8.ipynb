task 1.
Simulate a Bernoulli trial in Python to model whether a user successfully logs into their Zomato account (success=1, failure=0) with a given probability of success p=0.85. Run the simulation 20 times and print the results.


```python
import numpy as np
```

```python
p = 0.85

results = np.random.binomial(1, p, 20)

print("Zomato Login Simulation (20 Trials):")
print(results)

print("Successful Logins:", np.sum(results == 1))
print("Failed Logins:", np.sum(results == 0))
```

task 2.
Write a Python function that uses the binomial distribution to calculate the probability of getting exactly 3 successful UPI payments out of 5 attempts on Paytm, if the success rate per payment is 0.9.<br><br><em><strong>Hint:</strong> Use scipy.stats.binom.pmf for the calculation.</em>


```python
from scipy.stats import binom

# Given values
n = 5       # Total payment attempts
k = 3       # Exactly 3 successful payments
p = 0.9     # Probability of payment success

# Calculate binomial probability
probability = binom.pmf(k, n, p)

# Print result
print("Probability of exactly 3 successful UPI payments:", probability)
print("Probability in percentage:", probability * 100, "%")
```

task 3.
Suppose the average number of new playlist creations per minute on Spotify is 4. Use the Poisson distribution to calculate the probability that exactly 6 playlists are created in a given minute.<br><br><em><strong>Hint:</strong> Use scipy.stats.poisson.pmf.</em>


```python
from scipy.stats import poisson

lam = 4    # Average playlists created per minute
k = 6      # Exactly 6 playlists

probability = poisson.pmf(k, lam)

print("Probability of exactly 6 playlists:", probability)
print("Probability in percentage:", probability * 100, "%")
```

task 4.
Given a scenario where Flipkart receives an average of 10 product reviews per hour, write code to simulate and plot the number of reviews received in 12 randomly selected hours using the Poisson distribution.\<br>\<br>\<em>\<strong>Hint:\</strong> Use numpy.random.poisson and matplotlib for plotting.\</em>

```python
import numpy as np
import matplotlib.pyplot as plt

lam = 10

reviews = np.random.poisson(lam, 12)

print("Product Reviews in 12 Hours:")
print(reviews)

hours = np.arange(1, 13)

plt.figure(figsize=(10, 5))
plt.bar(hours, reviews)

plt.title("Flipkart Product Reviews per Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Reviews")
plt.xticks(hours)
plt.grid(axis='y')

plt.show()
```
