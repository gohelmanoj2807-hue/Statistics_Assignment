task 1.
Using Python and the scipy.stats library, perform a one-sample t-test to check if the average delivery time for Zomato orders in your city is significantly different from 30 minutes. Assume you have this sample data: [28, 32, 27, 35, 29, 31, 30, 33, 26, 34].


```python
import numpy as np
from scipy import stats

delivery_time = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]

population_mean = 30

t_stat, p_value = stats.ttest_1samp(delivery_time, population_mean)

sample_mean = np.mean(delivery_time)
print("Sample Mean:", sample_mean)
print("Population Mean:", population_mean)
print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject Null Hypothesis (H0)")
    print("Average delivery time is significantly different from 30 minutes.")
else:
    print("Fail to Reject Null Hypothesis (H0)")
    print("Average delivery time is not significantly different from 30 minutes.")
```

task 2.
Collect or simulate two independent samples of daily step counts: one group for users before installing a fitness app (e.g., Google Fit) and one group after installation. Use Python to perform an independent (two-sample) t-test and determine if the average steps per day increased after using the app.<br><br><em><strong>Hint:</strong> Use scipy.stats.ttest_ind().</em>


```python
before = [4500, 5200, 4800, 6100, 5300,
          4700, 5500, 4900, 5800, 5100]

after = [6200, 7000, 6500, 7200, 6800,
         6100, 7500, 6900, 6400, 7100]

print("Average steps before:", np.mean(before))
print("Average steps after:", np.mean(after))

t_stat, p_value = stats.ttest_ind(
    after, before, equal_var=False
)

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", p_value)

if p_value < 0.05:
    print("Reject Null Hypothesis (H0)")
    print("Daily steps increased significantly after using the app.")
else:
    print("Fail to Reject Null Hypothesis (H0)")
    print("No significant increase in daily steps.")
```

task 3.
Simulate paired data for 10 users showing their average screen time (in hours) before and after enabling 'Focus Mode' on their phones. Use Python to conduct a paired t-test and report whether 'Focus Mode' significantly reduced screen time.<br><br><em><strong>Hint:</strong> Use scipy.stats.ttest_rel().</em>


```python
before = [6.5, 7.0, 5.5, 8.0, 6.0,
          7.5, 6.8, 5.8, 7.2, 6.3]

after = [5.0, 5.5, 4.5, 6.0, 5.0,
         6.0, 5.2, 4.8, 5.5, 5.0]

# Step 2: Calculate average screen time
print("Average screen time before:", np.mean(before))
print("Average screen time after:", np.mean(after))

t_stat, p_value = stats.ttest_rel(before, after)

print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject Null Hypothesis (H0)")
    print("Focus Mode significantly reduced screen time.")
else:
    print("Fail to Reject Null Hypothesis (H0)")
    print("No significant reduction in screen time.")
```

task 4.
Given this scenario: You want to compare the average ratings of two food delivery apps (Zomato vs Swiggy) based on 15 randomly selected user ratings for each app. Write Python code to perform the appropriate t-test, and explain why you chose that test.


```python
zomato_ratings = [4.2, 4.5, 3.8, 4.0, 4.3,
                  4.1, 3.9, 4.6, 4.4, 4.0,
                  4.2, 3.7, 4.5, 4.1, 4.3]

swiggy_ratings = [4.0, 3.8, 4.1, 3.9, 4.2,
                  3.7, 4.0, 4.3, 3.9, 4.1,
                  3.8, 4.2, 4.0, 3.9, 4.1]

print("Zomato Average:", np.mean(zomato_ratings))
print("Swiggy Average:", np.mean(swiggy_ratings))

t_stat, p_value = stats.ttest_ind(
    zomato_ratings,
    swiggy_ratings,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant difference in average ratings.")
else:
    print("There is no significant difference in average ratings.")
```
