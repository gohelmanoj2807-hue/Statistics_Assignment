task 3.
Write a Python function bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email) that uses Bayes’ Theorem to compute the probability a user will churn given they received a 'we miss you' email. Test your function with prior_churn=0.15, prob_email_given_churn=0.6, prob_email=0.25.

```python
def bayes_churn_probability(prior_churn, prob_email_given_churn, prob_email):

    probability = (prior_churn * prob_email_given_churn) / prob_email

    return probability


# Test the function
result = bayes_churn_probability(
    prior_churn=0.15,
    prob_email_given_churn=0.6,
    prob_email=0.25
)

print("Probability of customer churn:", result)
print("Churn Probability (%):", result * 100)
```

A digital wallet app detects fraud based on spending patterns. If 2% of all transactions are fraudulent, and the app’s alert system correctly flags 90% of fraudulent transactions but also falsely flags 5% of normal transactions, use Bayes’ Theorem to calculate the probability that a transaction is actually fraudulent when flagged by the alert system.\<br>\<br>\<em>\<strong>Hint:\</strong> Define events: F = fraud, A = alert. Use Bayes’ formula: P(F|A) = [P(A|F) \* P(F)] / P(A).\</em>

Step 1: Calculate P(A)

P(A) = P(A|F) × P(F) + P(A|Not F) × P(Not F)

P(A) = (0.90 × 0.02) + (0.05 × 0.98)

P(A) = 0.018 + 0.049

P(A) = 0.067

Step 2: Apply Bayes' Theorem

P(A|B) = \frac{P(B|A)P(A)}{P(B)}

P(F|A) = (0.90 × 0.02) / 0.067

P(F|A) = 0.018 / 0.067

P(F|A) = 0.2687

Step 3: Convert into Percentage

0.2687 × 100 = 26.87%
