task 1.
Create a contingency table in Excel or Google Sheets showing the relationship between users' favorite food delivery app (Zomato, Swiggy, Uber Eats) and their preferred payment method (Paytm, PhonePe, Credit Card) using a sample of at least 20 responses.


```python
import pandas as pd
```

```python
data = {
    'Favorite_App': [
        'Zomato', 'Swiggy', 'Uber Eats', 'Zomato',
        'Swiggy', 'Uber Eats', 'Zomato', 'Swiggy',
        'Zomato', 'Uber Eats', 'Swiggy', 'Zomato',
        'Uber Eats', 'Swiggy', 'Zomato', 'Swiggy',
        'Uber Eats', 'Zomato', 'Swiggy', 'Uber Eats'
    ],

    'Payment_Method': [
        'Paytm', 'PhonePe', 'Credit Card', 'PhonePe',
        'Paytm', 'PhonePe', 'Credit Card', 'PhonePe',
        'Paytm', 'Credit Card', 'Credit Card', 'PhonePe',
        'Paytm', 'PhonePe', 'Credit Card', 'Paytm',
        'Credit Card', 'Paytm', 'Credit Card', 'PhonePe'
    ]
}

df = pd.DataFrame(data)

contingency_table = pd.crosstab(
    df['Favorite_App'],
    df['Payment_Method'],
    margins=True
)

print("Contingency Table:")
print(contingency_table)
```

task 2.
Given the following observed counts of music streaming preference among males and females: Spotify (Males: 12, Females: 18), Gaana (Males: 8, Females: 7), YouTube Music (Males: 10, Females: 15), calculate the expected counts for each cell assuming independence.<br><br><em><strong>Hint:</strong> Use the formula: (row total × column total) / grand total.</em>


```python
data = {
    'Music_App': ['Spotify', 'Gaana', 'YouTube Music'],
    'Males': [12, 8, 10],
    'Females': [18, 7, 15]
}

df = pd.DataFrame(data)

df['Row_Total'] = df['Males'] + df['Females']

male_total = df['Males'].sum()
female_total = df['Females'].sum()
grand_total = df['Row_Total'].sum()

df['Expected_Males'] = (
    df['Row_Total'] * male_total / grand_total
)

df['Expected_Females'] = (
    df['Row_Total'] * female_total / grand_total
)

print("Expected Counts:")
print(df[['Music_App', 'Expected_Males', 'Expected_Females']].round(2))
```

task 3.
Perform a chi-square test of independence using the data from your contingency table in task 1 to determine if there is a significant association between app choice and payment method. Show your calculation steps and state your conclusion.


```python
import numpy as np
from scipy.stats import chi2_contingency

# Observed data
observed = np.array([
    [3, 2, 2],  # Zomato
    [2, 3, 2],  # Swiggy
    [1, 2, 3]   # Uber Eats
])

chi2, p_value, dof, expected = chi2_contingency(
    observed, correction=False
)

print("Observed Table:")
print(observed)

print("\nExpected Counts:")
print(expected.round(2))

print("\nChi-Square Value:", round(chi2, 4))
print("P-value:", round(p_value, 4))
print("Degrees of Freedom:", dof)

if p_value < 0.05:
    print("\nReject the Null Hypothesis.")
    print("There is a significant association.")
else:
    print("\nFail to Reject the Null Hypothesis.")
    print("There is no significant association.")
```

task 4.
For a new Flipkart product launch, suppose you surveyed 30 users and observed the following: 18 prefer the new product, 12 do not. If the company claims that 70% of users will prefer the new product, perform a chi-square goodness of fit test to check this claim.<br><br><em><strong>Hint:</strong> Calculate expected counts based on the claimed proportion.</em>


```python
from scipy.stats import chisquare

observed = [18, 12]

expected = [21, 9]

chi2, p_value = chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Observed Counts:", observed)
print("Expected Counts:", expected)

print("Chi-Square Value:", round(chi2, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the Null Hypothesis.")
    print("The company's 70% claim is not supported.")
else:
    print("Fail to Reject the Null Hypothesis.")
    print("The company's 70% claim is not rejected.")
```
