task 1.
Download a small dataset of 10-15 songs with features like danceability, energy, and popularity (you can create a CSV manually or use any open dataset), then calculate the Pearson correlation coefficient between danceability and popularity using pandas.


```python
import pandas as pd

data = {
    'song': [
        'Song 1', 'Song 2', 'Song 3', 'Song 4',
        'Song 5', 'Song 6', 'Song 7', 'Song 8',
        'Song 9', 'Song 10', 'Song 11', 'Song 12'
    ],

    'danceability': [
        0.45, 0.62, 0.78, 0.55,
        0.81, 0.33, 0.70, 0.91,
        0.48, 0.67, 0.85, 0.39
    ],

    'energy': [
        0.50, 0.72, 0.88, 0.60,
        0.90, 0.40, 0.75, 0.95,
        0.52, 0.68, 0.86, 0.45
    ],

    'popularity': [
        35, 55, 82, 48,
        88, 25, 68, 95,
        40, 60, 90, 30
    ]
}

df = pd.DataFrame(data)

df.to_csv('songs.csv', index=False)

correlation = df['danceability'].corr(df['popularity'])

print("Pearson Correlation Coefficient:", correlation)
```

task 2.
Given a list of IPL players with their total runs and number of Instagram followers, use scipy.stats.spearmanr to calculate the Spearman correlation between runs and followers, and interpret whether higher runs generally mean more followers.


```python
from scipy.stats import spearmanr

data = {
    'player': [
        'Virat Kohli',
        'Rohit Sharma',
        'MS Dhoni',
        'KL Rahul',
        'Hardik Pandya',
        'Shubman Gill',
        'Suryakumar Yadav',
        'Rishabh Pant',
        'Jasprit Bumrah',
        'Ravindra Jadeja'
    ],

    'total_runs': [
        8000, 6500, 5200, 4500, 2500,
        3500, 3000, 4000, 1000, 2800
    ],

    'instagram_followers': [
        270000000, 45000000, 50000000,
        20000000, 40000000, 15000000,
        12000000, 10000000, 18000000, 8000000
    ]
}
df = pd.DataFrame(data)

correlation, p_value = spearmanr(
    df['total_runs'],
    df['instagram_followers']
)

print("Spearman Correlation:", correlation)
print("P-Value:", p_value)

if correlation > 0:
    print("Positive correlation: Higher runs generally mean more followers.")
elif correlation < 0:
    print("Negative correlation: Higher runs generally mean fewer followers.")
else:
    print("No correlation between runs and followers.")
```

task 3.
Create a correlation matrix for three Flipkart product features: price, user rating, and number of reviews, using pandas' corr() method, and display the result.


```python
data = {
    'price': [500, 1000, 1500, 2000, 2500, 3000],
    'user_rating': [3.5, 4.0, 4.2, 4.5, 4.7, 4.8],
    'number_of_reviews': [50, 100, 150, 250, 300, 400]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

task 4.
Visualize the correlation matrix you created in the previous task as a heatmap using seaborn's heatmap() function, and customize it by adding annotations and a color bar.<br><br><em><strong>Hint:</strong> Use sns.heatmap(corr_matrix, annot=True, cmap='coolwarm').</em>


```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'price': [500, 1000, 1500, 2000, 2500, 3000],
    'user_rating': [3.5, 4.0, 4.2, 4.5, 4.7, 4.8],
    'number_of_reviews': [50, 100, 150, 250, 300, 400]
}

df = pd.DataFrame(data)

corr_matrix = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    cbar=True,
    fmt='.2f'
)

plt.title('Flipkart Product Features Correlation Heatmap')

plt.show()
```

task 5.
List two potential pitfalls when interpreting correlation coefficients in data from real apps like Zomato or YouTube, and give a short example for each showing how correlation can be misleading.


(answer)

1. Correlation Does Not Mean Causation

A correlation between two variables does not mean that one causes the other.

Example: On YouTube, videos with more views may also have more likes. But more views do not necessarily cause more likes. Both may increase because the video is popular.

2. Influence of a Third Variable

Sometimes, another hidden factor affects both variables and creates a misleading correlation.

Example: On Zomato, restaurants with higher ratings may have more orders. However, restaurant location or advertising may also influence the number of orders.

Conclusion

Correlation shows a relationship between two variables, but it does not always explain why the relationship exists or prove that one variable causes another.
