task 1.
State the null and alternative hypotheses for testing whether a new cashback offer on a food delivery app (like Zomato) has increased the average number of daily orders compared to before the offer.


(answer)
Scenario: A food delivery app like Zomato introduces a new cashback offer to increase the average number of daily orders.

1. Null Hypothesis (H₀)

The new cashback offer does not increase the average number of daily orders.

H₀: μ_after ≤ μ_before

2. Alternative Hypothesis (H₁)

The new cashback offer increases the average number of daily orders.

H₁: μ_after > μ_before

Conclusion

We use hypothesis testing to check whether the cashback offer has significantly increased daily orders. If the p-value is less than 0.05, we reject the null hypothesis and conclude that there is sufficient statistical evidence that the cashback offer increased average daily orders.

task 2.
Given the following scenario: A music streaming app claims their new playlist recommendation algorithm increases average listening time from 40 to 45 minutes. If you set up a hypothesis test, identify which type of error (Type I or Type II) would occur if you wrongly conclude the algorithm works when it actually doesn't, and explain in one sentence why.


Answer: Type I Error (False Positive).

Explanation: A Type I error occurs when we wrongly reject the null hypothesis and conclude that the new playlist recommendation algorithm increases listening time, even though it actually does not.

task 3.
Suppose you run an A/B test on a shopping app (like Flipkart) and get a p-value of 0.03 when testing if a new discount banner increases purchases. At a 95% confidence level, state whether you reject or fail to reject the null hypothesis and explain what this means in the context of the app.


Given:

P-value = 0.03
Confidence level = 95%
Significance level (α) = 0.05

Decision:

Since the p-value (0.03) is less than the significance level (0.05), we reject the null hypothesis (H₀).

Explanation:

This means there is sufficient statistical evidence that the new discount banner increases purchases on Flipkart. Therefore, the new banner may help improve sales.

Conclusion: The new discount banner has a statistically significant positive effect on purchases at the 95% confidence level.

task 4.
For a cricket app's push notification feature, you want to be 99% confident that sending match alerts increases user engagement. Explain what the 99% confidence level means in this context, and how it affects your decision to accept or reject the feature.


Answer:

A 99% confidence level means we use a very strict statistical standard to check whether match alerts increase user engagement. It corresponds to a significance level of 0.01 (1%).

Decision:

If the p-value is less than 0.01, we reject the null hypothesis (H₀). There is sufficient evidence that match alerts increase user engagement.
If the p-value is greater than or equal to 0.01, we fail to reject the null hypothesis.

Conclusion:

A 99% confidence level helps the cricket app make more reliable decisions before launching the push notification feature to all users.
