task 2.
Given this list of food delivery times in minutes for your last 9 Zomato/Swiggy orders: [32, 28, 29, 45, 30, 31, 60, 30, 29], write a function to find the median delivery time.


```python
delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

def find_median(times):
    sorted_times = sorted(times)
    n = len(sorted_times)
    middle = n // 2

    if n % 2 == 1:
        return sorted_times[middle]
    else:
        return (sorted_times[middle - 1] + sorted_times[middle]) / 2

median_time = find_median(delivery_times)

print("Delivery Times:", delivery_times)
print("Sorted Times:", sorted(delivery_times))
print("Median Delivery Time:", median_time, "minutes")
```

task 3.
Imagine you tracked the genres of 12 YouTube videos you watched this week: ['Music', 'Vlog', 'Music', 'Tech', 'Music', 'Vlog', 'Tech', 'Music', 'Comedy', 'Vlog', 'Music', 'Comedy']. Write code to determine the mode (most watched genre).


```python
from collections import Counter

genres = [
    'Music', 'Vlog', 'Music', 'Tech',
    'Music', 'Vlog', 'Tech', 'Music',
    'Comedy', 'Vlog', 'Music', 'Comedy'
]

genre_count = Counter(genres)

mode_genre = genre_count.most_common(1)[0]

print("Video Genres:", genres)
print("Genre Frequency:", genre_count)
print("Most Watched Genre:", mode_genre[0])
print("Number of Views:", mode_genre[1])
```

task 4.
Take this array of daily UPI transaction amounts (in ₹): [100, 120, 105, 110, 5000, 115, 108]. Calculate both the mean and the median, then explain in 2-3 lines which is a better measure of your 'typical' transaction and why.<br><br><em><strong>Hint:</strong> Consider the effect of the ₹5000 value.</em>


```python
import numpy as np

transactions = np.array([100, 120, 105, 110, 5000, 115, 108])

mean_value = np.mean(transactions)
median_value = np.median(transactions)

print("UPI Transactions:", transactions)
print("Mean Transaction Amount: ₹", mean_value)
print("Median Transaction Amount: ₹", median_value)
```
