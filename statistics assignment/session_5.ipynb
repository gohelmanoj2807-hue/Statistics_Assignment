task 1.
Given a list of delivery times (in minutes) for 20 Zomato orders, calculate the 25th, 50th, and 75th percentiles using Python's numpy.percentile function.


```python
import numpy as np
```

```python
delivery_times = np.array([
    32, 28, 29, 45, 30, 31, 60, 30, 29, 35,
    40, 33, 27, 38, 42, 31, 34, 29, 50, 36
])

p25 = np.percentile(delivery_times, 25)
p50 = np.percentile(delivery_times, 50)
p75 = np.percentile(delivery_times, 75)

print("Delivery Times:", delivery_times)
print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

Create a Python script that takes an array of IPL player scores and divides them into quartiles using numpy.quantile. Print the values for Q1, Q2 (median), and Q3.

Given a list of daily step counts from your fitness app for one month, calculate the Interquartile Range (IQR) in Python and explain what it tells you about your activity consistency.<br><br><em><strong>Hint:</strong> IQR = Q3 - Q1</em>


```python
steps = np.array([
    6500, 7000, 7200, 6800, 7500,
    8000, 7100, 6900, 7300, 7600,
    7800, 7200, 7000, 7400, 8100,
    7700, 6900, 7200, 7500, 7300,
    6800, 7900, 8200, 7100, 7400,
    7600, 7000, 7300, 7800, 7500
])

q1 = np.percentile(steps, 25)
q3 = np.percentile(steps, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("Interquartile Range (IQR):", iqr)
```

task 4.
Write a Python function to detect outliers in a list of Swiggy order amounts using the IQR method (any value below Q1 - 1.5*IQR or above Q3 + 1.5*IQR). Print the outlier values.


```python
orders = np.array([
    250, 300, 280, 350, 400, 320,
    290, 310, 275, 330, 360, 295,
    305, 340, 280, 315, 300, 1200
])

def detect_outliers(data):
    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = data[(data < lower_limit) | (data > upper_limit)]

    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", outliers)

detect_outliers(orders)
```

task 5.
Using matplotlib, create a boxplot of your weekly screen time (in hours) for the past 8 weeks. Label the axes and highlight any outliers.


```python
import matplotlib.pyplot as plt
import numpy as np

screen_time = [3.5, 4.0, 3.8, 4.2, 3.9, 4.1, 4.0, 9.5]

q1 = np.percentile(screen_time, 25)
q3 = np.percentile(screen_time, 75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

outliers = [
    x for x in screen_time
    if x < lower_limit or x > upper_limit
]

plt.figure(figsize=(8, 5))

plt.boxplot(
    screen_time,
    patch_artist=True,
    flierprops=dict(
        marker='o',
        markerfacecolor='red',
        markersize=10,
        markeredgecolor='black'
    )
)

plt.title("Weekly Screen Time Boxplot")
plt.ylabel("Screen Time (Hours)")
plt.xlabel("8 Weeks")

for value in outliers:
    plt.annotate(
        f"Outlier: {value} hours",
        xy=(1, value),
        xytext=(1.15, value),
        arrowprops=dict(arrowstyle="->")
    )

plt.grid(axis='y')
plt.show()

print("Outliers:", outliers)
```
