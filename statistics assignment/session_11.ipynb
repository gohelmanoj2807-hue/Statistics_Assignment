task 1.
Simulate the Central Limit Theorem in Python by taking 1000 random samples (size 30) from a skewed dataset (e.g., number of likes on Instagram posts), calculating the mean of each sample, and plotting the histogram of these sample means using matplotlib.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

likes = np.random.exponential(scale=500, size=10000)

sample_means = []

for i in range(1000):
    sample = np.random.choice(likes, size=30)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)

plt.figure(figsize=(10, 5))

plt.hist(sample_means, bins=30, edgecolor='black')

plt.title('Central Limit Theorem - Sampling Distribution')
plt.xlabel('Sample Means (Instagram Likes)')
plt.ylabel('Frequency')

plt.show()

# Step 4: Display results
print("Total Samples:", 1000)
print("Sample Size:", 30)
print("Mean of Sample Means:", np.mean(sample_means))
print("Original Dataset Mean:", np.mean(likes))
```

task 2.
Explain in your own words why the sampling distribution of the mean becomes approximately normal, even if the original data is not, using an example from any app you use daily (such as Zomato order amounts or Spotify song durations).


(amswer)Task 2: Explanation of Central Limit Theorem (CLT)

The sampling distribution of the mean becomes approximately normal because when we take many random samples from a population and calculate their averages, the sample means tend to form a bell-shaped curve.

This happens even if the original data is skewed or not normally distributed, provided the sample size is sufficiently large.

Example: Zomato Order Amounts

Suppose we collect data about 10,000 Zomato orders. Some customers order food worth ₹100, while others order ₹500 or ₹1,000. Therefore, the original data may be skewed.

Now:

1.Take 100 random samples, each containing 30 orders.
2.Calculate the average order amount for each sample.
3.Repeat this process many times.
4.Plot a histogram of all the sample averages.

The histogram will look approximately like a bell-shaped normal distribution.

Conclusion:

The Central Limit Theorem tells us that the distribution of sample means becomes approximately normal when we take sufficiently large random samples, even if the original data is not normally distributed. It is very useful in data analytics for making predictions and statistical decisions.

task 4.
Modify the CLT simulation code to use a much smaller sample size (n=5) and compare the histogram of sample means to the one with n=30. Briefly describe what changes you observe in the shape of the distribution.<br><br><em><strong>Hint:</strong> Run both simulations and compare the spread and normality of the histograms.</em>


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

likes = np.random.exponential(scale=500, size=10000)

sample_means_5 = []

for i in range(1000):
    sample = np.random.choice(likes, size=5)
    sample_means_5.append(np.mean(sample))

sample_means_30 = []

for i in range(1000):
    sample = np.random.choice(likes, size=30)
    sample_means_30.append(np.mean(sample))

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(sample_means_5, bins=30, edgecolor='black')
plt.title('Sample Means (n=5)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(sample_means_30, bins=30, edgecolor='black')
plt.title('Sample Means (n=30)')
plt.xlabel('Sample Mean')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

print("Mean for n=5:", np.mean(sample_means_5))
print("Mean for n=30:", np.mean(sample_means_30))
print("Standard Deviation for n=5:", np.std(sample_means_5))
print("Standard Deviation for n=30:", np.std(sample_means_30))
```

task 5.
Use ChatGPT to generate Python code that demonstrates the Central Limit Theorem using random samples from a non-normal distribution (e.g., exponential). Run the code, paste the output plot, and briefly describe what you observe.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

population = np.random.exponential(scale=500, size=10000)

sample_means = []

for i in range(1000):
    sample = np.random.choice(population, size=30, replace=True)
    sample_means.append(np.mean(sample))

plt.figure(figsize=(9, 5))

plt.hist(sample_means, bins=30, edgecolor='black')

plt.title('Central Limit Theorem: Sample Means (n=30)')
plt.xlabel('Sample Mean of Instagram Likes')
plt.ylabel('Frequency')

plt.show()

print("Number of samples:", 1000)
print("Sample size:", 30)
print("Mean of sample means:", round(np.mean(sample_means), 2))
print("Population mean:", round(np.mean(population), 2))
```
