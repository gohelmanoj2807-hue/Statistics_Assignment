task 2.
Given an array of 1000 Flipkart orders labeled by city (e.g., 'Ahmedabad', 'Surat', 'Vadodara'), write code to generate a stratified sample of 60 orders, ensuring each city is proportionally represented in the sample.<br><br><em><strong>Hint:</strong> Calculate the proportion of orders per city and sample accordingly from each group.</em>


```python
import pandas as pd
import numpy as np

np.random.seed(42)

cities = np.random.choice(
    ['Ahmedabad', 'Surat', 'Vadodara'],
    size=1000,
    p=[0.5, 0.3, 0.2]
)

orders_df = pd.DataFrame({
    'Order_ID': range(1, 1001),
    'City': cities
})

city_proportions = orders_df['City'].value_counts(normalize=True)

sampled_orders = (
    orders_df.groupby('City', group_keys=False)
    .apply(lambda group: group.sample(
        n=round(len(group) * 60 / 1000),
        random_state=42
    ))
)

print("Stratified Sample of 60 Orders:")
print(sampled_orders)

print("\nOrders Selected From Each City:")
print(orders_df.loc[sampled_orders.index, 'City'].value_counts())

print("\nTotal Sample Size:", len(sampled_orders))
```

task 3.
Write a function to perform systematic sampling on a list of 500 Zomato restaurant reviews, selecting every 10th review to create a sample set.


```python
def systematic_sampling(reviews, interval=10):

    sample = reviews[::interval]

    return sample


reviews = [f"Review {i}" for i in range(1, 501)]

sample_reviews = systematic_sampling(reviews)

print("Systematically Selected Reviews:")
print(sample_reviews)

print("\nTotal Reviews:", len(reviews))
print("Total Sampled Reviews:", len(sample_reviews))
```

task 5.
Modify the stratified sampling demo shown in class: instead of customer segments, use a dataset of IPL cricket fans grouped by favorite team, and generate a stratified sample of 40 fans for a survey, ensuring each team is represented proportionally.<br><br><em><strong>Constraint:</strong> Do not copy the code from the demo—write your own logic for grouping and sampling.</em>


```python
import pandas as pd
import random

random.seed(42)

teams = (
    ['CSK'] * 50 +
    ['MI'] * 40 +
    ['RCB'] * 35 +
    ['KKR'] * 30 +
    ['GT'] * 25 +
    ['RR'] * 20
)

fans = pd.DataFrame({
    'Fan_ID': range(1, 201),
    'Favorite_Team': teams
})

fans = fans.sample(frac=1, random_state=42).reset_index(drop=True)

team_groups = {}

for team in fans['Favorite_Team'].unique():
    team_groups[team] = fans[fans['Favorite_Team'] == team]

stratified_sample = []

for team, group in team_groups.items():

    proportion = len(group) / len(fans)
    sample_count = round(proportion * 40)

    selected = group.sample(
        n=sample_count,
        random_state=42
    )

    stratified_sample.append(selected)

# Combine all selected fans
final_sample = pd.concat(stratified_sample)

# Display results
print("Stratified Sample of IPL Fans:")
print(final_sample.sort_values('Fan_ID'))

print("\nTeam-wise Sample Count:")
print(final_sample['Favorite_Team'].value_counts())

print("\nTotal Selected Fans:", len(final_sample))
```
