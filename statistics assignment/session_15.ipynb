task 1.
Create a CSV file named purchase_data.csv with columns: customer_id, age_group (18-25, 26-35, 36-50), and purchase_value. Enter at least 15 sample records with 5 customers in each age group.


```python
import pandas as pd

data = {
    'customer_id': range(1, 16),

    'age_group': [
        '18-25', '18-25', '18-25', '18-25', '18-25',
        '26-35', '26-35', '26-35', '26-35', '26-35',
        '36-50', '36-50', '36-50', '36-50', '36-50'
    ],

    'purchase_value': [
        450, 520, 390, 610, 480,
        750, 820, 690, 900, 780,
        1100, 1250, 980, 1350, 1180
    ]
}

df = pd.DataFrame(data)

df.to_csv('purchase_data.csv', index=False)

print(df)

print("CSV file created successfully!")
```

task 2.
Write a Python script using pandas and scipy.stats that reads purchase_data.csv and performs a one-way ANOVA to compare the average purchase_value across the three age groups. Print the F-statistic and p-value.


```python
import pandas as pd
from scipy.stats import f_oneway

df = pd.read_csv('purchase_data.csv')

group1 = df[df['age_group'] == '18-25']['purchase_value']
group2 = df[df['age_group'] == '26-35']['purchase_value']
group3 = df[df['age_group'] == '36-50']['purchase_value']

f_stat, p_value = f_oneway(group1, group2, group3)

print("One-Way ANOVA Results")
print("F-Statistic:", f_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("There is a significant difference between the group averages.")
else:
    print("There is no significant difference between the group averages.")
```

task 4.
Calculate and print the between-group variance and within-group variance for the purchase_value data using numpy or pandas. Explain briefly what each value represents in this context.<br><br><em><strong>Hint:</strong> Between-group variance measures differences between age group means; within-group variance measures variation inside each age group.</em>


```python
import pandas as pd
import numpy as np

df = pd.read_csv('purchase_data.csv')

grand_mean = df['purchase_value'].mean()

groups = df.groupby('age_group')['purchase_value']

ss_between = sum(
    len(group) * (group.mean() - grand_mean) ** 2
    for name, group in groups
)

df_between = df['age_group'].nunique() - 1
between_variance = ss_between / df_between

ss_within = sum(
    ((group - group.mean()) ** 2).sum()
    for name, group in groups
)

df_within = len(df) - df['age_group'].nunique()
within_variance = ss_within / df_within

print("Between-Group Variance:", between_variance)
print("Within-Group Variance:", within_variance)
```

task 5.
Imagine you are analyzing Zomato order data for three different customer segments: Students, Working Professionals, and Retirees. Use ChatGPT or Copilot to generate Python code that performs a one-way ANOVA on their average monthly order amounts. Paste the generated code and run it with dummy data for each group.


```python
import pandas as pd
from scipy.stats import f_oneway

students = [800, 950, 700, 1100, 900]

working_professionals = [1800, 2000, 1600, 2200, 1900]

retirees = [1200, 1300, 1100, 1400, 1250]

print("Average Monthly Order Amounts:")
print("Students:", sum(students) / len(students))
print("Working Professionals:", sum(working_professionals) / len(working_professionals))
print("Retirees:", sum(retirees) / len(retirees))

f_stat, p_value = f_oneway(
    students,
    working_professionals,
    retirees
)

print("\nOne-Way ANOVA Results")
print("F-Statistic:", f_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("There is a significant difference between customer segments.")
else:
    print("There is no significant difference between customer segments.")
```
