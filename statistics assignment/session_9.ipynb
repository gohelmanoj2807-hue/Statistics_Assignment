task 1.
Write a Python script that generates 1000 random numbers following a normal distribution with mean 500 and standard deviation 100, then plot the resulting distribution using matplotlib.


```python
import numpy as np
import matplotlib.pyplot as plt

mean = 500
std_dev = 100

data = np.random.normal(mean, std_dev, 1000)

plt.figure(figsize=(10, 5))
plt.hist(data, bins=30, edgecolor='black')

plt.title("Normal Distribution of Sales Amounts")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

plt.grid(axis='y')
plt.show()
```

task 2.
Given an array of daily step counts from a fitness app, calculate the Z-score for a specific day where the user walked 9500 steps. Assume the mean is 8000 and standard deviation is 1200.<br><br><em><strong>Hint:</strong> Use the formula Z = (X - μ) / σ.</em>


```python
X = 9500       # Steps walked
mean = 8000    # Average steps
std_dev = 1200 # Standard deviation

z_score = (X - mean) / std_dev

print("Z-Score:", z_score)
```

task 3.
Plot a standard normal curve (mean 0, standard deviation 1) using matplotlib, and shade the area between Z = -1 and Z = 1 to visualize the probability of values within one standard deviation.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mean = 0
std_dev = 1

x = np.linspace(-4, 4, 1000)

y = norm.pdf(x, mean, std_dev)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label="Standard Normal Curve")

x_fill = np.linspace(-1, 1, 500)
y_fill = norm.pdf(x_fill, mean, std_dev)

plt.fill_between(x_fill, y_fill, alpha=0.4, label="Area between Z = -1 and Z = 1")

plt.title("Standard Normal Distribution")
plt.xlabel("Z-Score")
plt.ylabel("Probability Density")
plt.axvline(-1, linestyle='--')
plt.axvline(1, linestyle='--')

plt.legend()
plt.grid()
plt.show()

probability = norm.cdf(1) - norm.cdf(-1)
print("Probability between Z = -1 and Z = 1:", probability)
print("Percentage:", probability * 100, "%")
```

task 4.
You are analyzing the delivery times (in minutes) for a food delivery app like Zomato. Given a normal distribution with mean 30 and standard deviation 5, write code to find the probability that a randomly selected delivery takes less than 25 minutes.<br><br><em><strong>Hint:</strong> Use scipy.stats.norm.cdf().</em>


```python
from scipy.stats import norm

mean = 30
std_dev = 5
delivery_time = 25

probability = norm.cdf(delivery_time, mean, std_dev)

print("Probability of delivery taking less than 25 minutes:", probability)
print("Probability in percentage:", probability * 100, "%")
```
