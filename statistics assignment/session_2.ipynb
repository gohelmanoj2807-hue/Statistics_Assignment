task 4.
Write a function or pseudocode that takes a column of data (array of values) from a Flipkart order history and automatically detects whether it is likely Nominal, Ordinal, Interval, or Ratio scale. Briefly describe the logic used for each type.<br><br><em><strong>Constraint:</strong> You do not need to write a perfect algorithm — focus on simple rules based on value types and patterns.</em>


```python
def detect_scale(data):

    data = [x for x in data if x is not None]

    if len(data) == 0:
        return "No Data"

    if all(isinstance(x, (int, float)) for x in data):

        if min(data) >= 0:
            return "Likely Ratio"

        else:
            return "Likely Interval"

    ordinal_values = [
        "Low", "Medium", "High",
        "Poor", "Average", "Good",
        "Excellent"
    ]

    if all(x in ordinal_values for x in data):
        return "Ordinal"

    return "Nominal"



payment_method = ["UPI", "Cash", "Card", "UPI"]

product_rating = ["Good", "Excellent", "Average"]

product_price = [500, 1000, 1500, 2000]

temperature = [-5, 0, 5, 10]


print("Payment Method:", detect_scale(payment_method))

print("Product Rating:", detect_scale(product_rating))

print("Product Price:", detect_scale(product_price))

print("Temperature:", detect_scale(temperature))
```
