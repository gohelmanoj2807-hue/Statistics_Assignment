List all possible outcomes (sample space) for rolling a standard six-sided die and flipping a coin at the same time. Write your answer as a list of ordered pairs.

task 2.
Suppose you have a deck of 10 cards numbered 1 to 10. Calculate the classical probability of drawing a card with an even number. Show your calculation steps.


```python
total_cards = 10
even_cards = 5

probability = even_cards / total_cards

print("Total Cards:", total_cards)
print("Even Cards:", even_cards)
print("Probability of Even Number:", probability)
print("Percentage:", probability * 100, "%")
```

task 3.
In a playlist of 8 songs on Spotify, 3 are by Arijit Singh. If you shuffle and play one random song, what is the probability that it is NOT by Arijit Singh? Show your working.


```python
total_songs = 8
arijit_songs = 3

other_songs = total_songs - arijit_songs

probability = other_songs / total_songs

print("Total Songs:", total_songs)
print("Songs Not by Arijit Singh:", other_songs)
print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

task 4.
A Zomato user places an order. The probability they order Pizza is 0.3, and the probability they order Burger is 0.4. If the probability they order both is 0.1, use the addition rule to find the probability the user orders Pizza or Burger.<br><br><em><strong>Hint:</strong> P(A or B) = P(A) + P(B) - P(A and B)</em>


```python
pizza = 0.3
burger = 0.4
both = 0.1

probability = pizza + burger - both

print("Probability of Pizza or Burger:", probability)
print("Percentage:", probability * 100, "%")
```

task 5.
A Flipkart user is browsing electronics. The probability they add a phone to cart is 0.2. If you know they are interested in mobiles, the probability rises to 0.5. Briefly explain in your own words what this conditional probability means, using this scenario.


task 5 (answer)

Explanation (in simple words):

Conditional probability means the probability of an event happening when we already know some information.

In this example, normally there is a 20% chance that a user adds a phone to the cart. But if we know that the user is interested in mobiles, the probability increases to 50%. This means interested customers are more likely to add a phone to their cart.
